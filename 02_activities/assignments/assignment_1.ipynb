### Part 1: Building the base Anagram Checker

Given two valid strings, check to see if they are anagrams of each other. If it is, return `True`, else `False`. For this part, we can assume that uppercase letters are the same as if it was a lowercase character.

Examples of anagrams:
* Silent and Listen
* Night and Think

Example outputs:
```python
anagram_checker("Silent", "listen") # True
anagram_checker("Silent", "Night") # False
anagram_checker("night", "Thing") # True
```

````python
# For testing purposes, we will write our code in the function
def anagram_checker(word_a, word_b):
    '''Test whether word_a and word_b are anagrams. word_a and word_b are strings. 
    We are not testing the case of the letters and so the words will first be converted to all uppercase.
    The strings will then be broken into indiviudal letters as lists of strings and then sorted alphabetically.
    The sorted lists will then be compared to determine if they are equivalent.
    True means that the words are anagrams, while false means that they are not.'''
    
    upper_word_a = word_a.upper()
    upper_word_b = word_b.upper()
    letters_a = list(upper_word_a)
    letters_b = list(upper_word_b)
    organized_a = sorted(letters_a)
    organized_b = sorted(letters_b)

    if organized_a == organized_b:
        return True
    else:
        return False




# Run your code to check using the words below:
anagram_checker("Silent", "listen")
````

````python
anagram_checker("Silent", "Night")
````

````python
anagram_checker("night", "Thing")
````

### Part 2: Expanding the functionality of the Anagram Checker

Using your existing and functional anagram checker, let's add a boolean option called `is_case_sensitive`, which will return `True` or `False` based on if the two compared words are anagrams and if we are checking for case sensitivity.

````python
def anagram_checker(word_a, word_b, is_case_sensitive):
  # Modify your existing code here
    '''This function will test whether the strings word_a and word_b are anagrams. 
    is_case_sensitive is a boolean that will determine whether we are considering the case of the letters in word_a and word_b.
    If testing with case sensitivity the strings will be converted to lists, sorted and compared.
    Otherwise the function will first make the strings all capitals and then convert to lists, sort and compare.
    True means that the words are anagrams, while false means that they are not and take into account case sensitivity depending on the input.'''

    if is_case_sensitive == True: 
        letters_a = list(word_a)
        letters_b = list(word_b)
        organized_a = sorted(letters_a)
        organized_b = sorted(letters_b)
        if organized_a == organized_b:
            return True
        else:
            return False

    else:
        upper_word_a = word_a.upper()
        upper_word_b = word_b.upper()
        letters_a = list(upper_word_a)
        letters_b = list(upper_word_b)
        organized_a = sorted(letters_a)
        organized_b = sorted(letters_b)

        if organized_a == organized_b:
            return True
        else:
            return False


# Run your code to check using the words below:
anagram_checker("Silent", "listen", False) # True
````

````python
anagram_checker("Silent", "Listen", True) # False
````
